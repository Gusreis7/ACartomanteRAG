```python
# importing required modules
from pypdf import PdfReader
import re
```

```python
def clean_text(text):
    """
    Limpar  quebras de linhas excessivas
    """
    
    pattern_duplo = re.compile('\n{2,}',re.S)
    text = re.sub(pattern_duplo, '\n\n',text)
    pattern = re.compile('\n \s{2,}', re.S)
    text = re.sub(pattern, '\n\n ', text)    


    return text


def split_capitulos(text):
    """
    Recebe o texto completo de uma obra e divide em capítulos.
    Funciona com títulos como 'CAPÍTULO PRIMEIRO' ou 'CAPÍTULO IV'.
    Retorna uma lista de capítulos.
    """
    pattern_capitulos = re.compile(
        r'(CAP[IÍ]TULO\s+(?:[A-ZÇÂÊÔÃÕ]+|\b[IVXLCDM]+\b))', re.MULTILINE
    )

    partes = pattern_capitulos.split(text)

    capitulos = []
    for i in range(1, len(partes), 2):
        titulo = partes[i].strip()
        conteudo = partes[i + 1].strip() if i + 1 < len(partes) else ""
        capitulos.append(f"{titulo}\n{conteudo}")

    return capitulos
```

```python
reader = PdfReader('data/Machado/romance/casaVelha.pdf')

text = ""
print(len(reader.pages))
```

```python

for page in reader.pages:
    text_page = page.extract_text()
    text_page = clean_text(text_page)
    text+=" "+text_page

print(len(text))
```

```python
capitulos = split_capitulos(text=text)
```
